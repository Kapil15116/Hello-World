```python
import tiktoken
```

```python
text= "PM of india is modiji"
```

```python
tokenizer =tiktoken.encoding_for_model(model_name="gpt-4")
#[10782, 328, 42045, 382, 1812, 15012]
```

```python
tokenIds =tokenizer.encode(text)
tokenIds
```

```python
resultIds =[8971, 315, 28811, 374, 1491, 35973]
```

```python
result =tokenizer.decode(resultIds)
result
```
